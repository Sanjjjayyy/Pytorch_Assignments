# Assignment 13 — LSTM Fundamentals


## Part A — Create Dataset

Create 

- X.shape -> (60, 8, 12)
- y.shape -> 0, 3

```python
import torch

X=torch.rand([60,8,12],dtype=torch.float32)
y=torch.randint(0,3,(60,))

print(X.shape)
print(y.shape)
```

## Part B — Build LSTM Model

Architecture:

- LSTM(
    input_size=12,
    hidden_size=20,
    batch_first=True
  )

- Linear(20,3)

```python
import torch.nn as nn

class LSTM_Model(nn.Module):
    def __init__(self):
        super().__init__()

        self.lstm=nn.LSTM(
            input_size=12,
            hidden_size=20,
            batch_first=True
        )

        self.linear=nn.Linear(20,3)

    def forward(self,X):

        output,(hidden,cell)=self.lstm(X)
        output=output[:,-1,:]
        output=self.linear(output)

        return output
    
model=LSTM_Model()
```

## Part D — Train

Use:

- CrossEntropyLoss
- Adam

Train:

- 150 epochs

```python
import torch.optim as optim

loss_fn=nn.CrossEntropyLoss()
optimizer=optim.Adam(model.parameters(),lr=0.01)


for epoch in range(150):
    
    predicitons=model(X)
    loss=loss_fn(predicitons,y)
    optimizer.zero_grad
    loss.backward()
    optimizer.step()

    if epoch%20==0:
        print(f"Epoch: {epoch} Train Loss :, {loss.item()}")

```

## Part F — Predictions

- Get predicted classes using:argmax

```python
with torch.no_grad():
    logits=model(X)
    preds=torch.argmax(logits,dim=1)
print(preds)
```
