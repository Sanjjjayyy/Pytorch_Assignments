# Assignment 13 — LSTM Fundamentals

## Part A — Create Dataset

- Create : X shape = (60, 8, 12)

- Classes : 0 or 1

```python
import torch
import torch.nn as nn

X=torch.rand([50,5,10],dtype=torch.float32)
y=torch.randint(0,2,(50,))

print(X)
print(y)
print(X.shape)
print(y.shape)
```

## Part B — Build RNN Model

Architecture:

- LSTM -> input_size=10, hidden_size=16, batch_first=True

- Linear(16,2)

```python
class RNN_Model(nn.Module):

    def __init__(self):
        super().__init__()

        self.rnn=nn.RNN(
            input_size=10,
            hidden_size=16,
            batch_first=True
        )
        self.linear=nn.Linear(16,2)

    def forward(self,X):

        output,hidden=self.rnn(X)
        output=output[:, -1, :]
        output = self.linear(output)

        return output
    
model=RNN_Model()
```

## Part C — Print Shapes

Inside forward print:

- RNN output shape
- Last timestep shape

```python
class RNN_Model_Print_Shapes(nn.Module):

    def __init__(self):
        super().__init__()

        self.rnn=nn.RNN(
            input_size=10,
            hidden_size=16,
            batch_first=True
        )
        self.linear=nn.Linear(16,2)


    def forward(self,X):

        output,hidden=self.rnn(X)
        print("RNN Output Shape:",output.shape)

        output=output[:,-1,:]
        print("Last Timestep Shape:",output.shape)
     
        output=self.linear(output)

        return output
    
print_shape=RNN_Model_Print_Shapes()
print_shape(X)
```

## Part D — Train

Use:

- CrossEntropyLoss
- Adam

Train:

- 150 epochs

```python
import torch.optim as optim

loss_fn=nn.CrossEntropyLoss()
optimizer=optim.Adam(model.parameters(),lr=0.01)


for epoch in range(200):
    predictions=model(X)
    loss=loss_fn(predictions,y)
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    if epoch%20==0:
        print(f"Epoch: {epoch} Train Loss :, {loss.item()}")
```

## Part E — Predictions

- Get predicted classes using:argmax

```python

with torch.no_grad():
    logits=model(X)
    preds=torch.argmax(logits, dim=1)
print(preds)
```
