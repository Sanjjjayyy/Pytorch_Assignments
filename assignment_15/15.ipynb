# Assignment 14 — GRU Fundamentals

## Part A — Create Dataset

- X.shape -> (80, 10, 15)
- y = (0,1,2,3)

```python
import torch

X=torch.rand([80,10,15],dtype=torch.float32)
y=torch.randint(0,4,(80,))

print(X)
print(y)

print(X.shape)
print(y.shape)
```

## Part B — Build GRU Model

Architecture:

- GRU(
    input_size=15,
    hidden_size=24,
    batch_first=True
  )

- Linear(24,4)

```python
import torch
import torch.nn as nn

class GRU_Model(nn.Module):

    def __init__(self):
        super().__init__()

        self.gru=nn.GRU(
            input_size=15,
            hidden_size=24,
            batch_first=True
        )
        self.linear=nn.Linear(24,4)

    def forward(self,X):
        output,hidden=self.gru(X)
        output=output[:,-1,:]
        output=self.linear(output)

        return output
    
model=GRU_Model()
```

## Part C — Train

Use:

- CrossEntropyLoss
- Adam

Train for:

- 150 epochs

```python
import torch.optim as optim

loss_fn=nn.CrossEntropyLoss()
optimizer=optim.Adam(model.parameters(),lr=0.01)


for epoch in range(150):
    
    predicitons=model(X)
    loss=loss_fn(predicitons,y)
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    if epoch%10==0:
        print(f"Epoch: {epoch} Train Loss :, {loss.item()}")

```

## Part F — Predictions

- Use : torch.argmax(logits, dim=1) to obtain class predictions.

```python
model.eval()

with torch.no_grad():
    logits=model(X)
    preds=torch.argmax(logits,dim=1)
print(preds)
```
