# Assignment 16 — Loss Functions

## Part A — Create a Regression Dataset

Create a synthetic regression dataset with the following requirements:

- Generate 1000 samples, each containing 5 input features drawn from a standard normal distribution.

Construct the target variable y as a linear combination of the first three features:

- Feature 1 should have a coefficient of 3
- Feature 2 should have a coefficient of −2
- Feature 3 should have a coefficient of 1.5
- Add random Gaussian noise to the target variable to simulate real-world data.
- Reshape the target variable so that it has dimensions (1000, 1).
- Verify the shapes of both the feature matrix and the target vector.

```python
import torch

# Features
X=torch.randn(1000,5)

# Target
y=(
    3*X[:,0]
    -2*X[:,1]
    +1.5*X[:,2]
    +0.5*torch.randn(1000)
)

# Converting y from (1000,) to (1000, 1)
y=y.view(-1,1)

print(X.shape)  
print(y.shape)  
```

## Part B — Build a Simple Regression Model

Use:

- nn.Linear(5, 1)

- No hidden layers.

```python
import torch
import torch.nn  as nn

class LinearModel(nn.Module):

    def __init__(self):

        super().__init__()
        self.linear=nn.Linear(5,1)

    def forward(self,x):

        x=self.linear(x)
        return x

model=LinearModel()
```

## Part C — Train with MSE Loss

Train for:

- 100 epochs
- Adam optimizer
- lr = 0.01

```python
import torch.optim as optim

loss_fn=nn.MSELoss()
optimizer=optim.Adam(model.parameters(),lr=0.01)
loss_history = []

for epoch in range(100):
    
    predicitons=model(X)
    loss=loss_fn(predicitons,y)
    # Storing loss hisrtory
    loss_history.append(loss.item())
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    if epoch%10==0:
        print(f"Epoch: {epoch} Train Loss :, {loss.item()}")

print(loss_history)

```

## Part D — Train with L1 Loss

- Same model architecture.

- Replace : nn.MSELoss() with L1Loss

```python
import torch.optim as optim

loss_fn=nn.L1Loss()
optimizer=optim.Adam(model.parameters(),lr=0.01)
loss_history = []

for epoch in range(100):
    
    predicitons=model(X)
    loss=loss_fn(predicitons,y)
    # Storing loss hisrtory
    loss_history.append(loss.item())
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    if epoch%10==0:
        print(f"Epoch: {epoch} Train Loss :, {loss.item()}")

print(loss_history)

```

## Part E — Train with SmoothL1 Loss

- Same experiment.
